```python
#Assignment
import pandas as pd

df = pd.read_csv("/content/archive.zip")
df.head()
```

```python
#1.Total Passengers ?
print(f"Total passengers are :{len(df["PassengerId"])}")
```

```python
#2.Average age of Survivors and non Survivors

print(f"Average age of Survivors: {df[df["Survived"]==1]["Age"].mean()}")
print(f"Average age of non survivors: {df[df["Survived"]==0]["Age"].mean()}")
```

```python
#3.List the Passengers above 60

print(f"Passengers above 60 are:")
for name in df[df['Age']>60]['Name']:
  print(name)
```

```python

#find the avg fare for each passenger class?
print("Average fare for each passenger class :")
print(df.groupby("Pclass")["Fare"].mean())
```
