```python
#importing libraries
import numpy as np
import pandas as pd
```

```python
#uploading Datasets

df =pd.read_csv('/content/score_updated.csv')
df.head()
```

```python
df.shape
```

```python
# Define input data
x = df[["Hours"]]

# Define target data
y = df["Scores"]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)


# Instantiate the LinearRegression model
decision_model = LinearRegression()

decision_model.fit(x_train, y_train)

decision_pred = decision_model.predict(x_test)
```

```python
# Predict using the Decision Tree model
decision_pred = decision_model.predict(x_test)

print(decision_pred)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Evaluation metrics
mse = mean_squared_error(y_test, decision_pred)
mae = mean_absolute_error(y_test, decision_pred)
rmse = mean_squared_error(y_test, decision_pred) ** 0.5
r2 = r2_score(y_test, decision_pred)

print("MSE :", mse)
print("MAE :", mae)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Flatten x_test.values to ensure sort_index is 1D
x_test_flat = x_test.values.flatten()
sort_index = np.argsort(x_test_flat)

plt.figure(figsize=(8, 5))

# Plot scatter of actual values, using flattened x_test
plt.scatter(x_test_flat, y_test, color="blue", label="Actual")

# Plot the regression line, using 1D sorted arrays
plt.plot(
    x_test_flat[sort_index],
    decision_pred[sort_index],
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.title("Linear Regression", fontsize=14, fontweight="bold")
plt.xlabel("Study hours")
plt.ylabel("Marks")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Take input from user
input_value = float(input("Enter the Hours Studied: "))

# Predict marks
prediction = decision_model.predict([[input_value]])

print("Predicted marks:", round(prediction[0], 0), "Marks")
```
