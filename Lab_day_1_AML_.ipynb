```python
1. #crating an array

import numpy as np
a = np.array([10,20,30,40,50])
print(a)

b = np.arange(5,11)
print(b)

c = np.linspace(1,10,5)
print(c)
```

```python
2.#matrix operations
a= np.array([[5,4],[10,20]])

b= np.array([[10,20],[30,40]])

print(a+b)
print(a-b)
print(a*b)
print(a@b)
```

```python
3. #Statastics

marks =np.array([20,30,50,80,90,94,99])

print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.std(marks))
```
